```python
import numpy as np
from bokeh.io import show, output_notebook
from bokeh.plotting import figure, gridplot

output_notebook()
```

```python
BATCH_SIZE = 20
```

```python
mean = np.array([3., 3.])
cov = np.array([[2., 1.5], [1.5, 2.]])
data = np.random.multivariate_normal(mean, cov, size=100)

fig = figure(
    title='Linear subspace',
    width=400,
    height=400,
    x_range=(-5, 10),
    y_range=(-5, 10),
)
fig.circle(data[:, 0], data[:, 1])
show(fig)

data_lin = data
data_lin_backup = np.copy(data_lin)
```

```python
%load_ext autoreload
%autoreload 2
from vae_lib import plot_loss, loss_function, VAE, run_training, NB_SAMPLES
```

```python
model_lin = VAE(latent_dim=2, nb_hidden=2)

from torch import optim
optim_lin = optim.Adam(model_lin.parameters(), lr=1e-3, weight_decay=1e-6)
elbos_lin = []
klds_lin = []
```

```python
run_training(data_lin, model_lin, optim_lin, 10000, elbos_lin, klds_lin, log_interval=500, batch_size=BATCH_SIZE)
```

```python
lin_box = ((-2,-3), (8,8))
fig1 = figure(
    title='p(x)',
    width=400,
    height=400,
    x_range=(lin_box[0][0], lin_box[1][0]),
    y_range=(lin_box[0][1], lin_box[1][1]),
)
plot_loss(model_lin, fig1, data_lin, nb_samples=NB_SAMPLES, box=lin_box, resolution=150)

fig2 = figure(
    title='ELBO Progress',
    width=400,
    height=400,
)
fig2.line(range(len(elbos_lin)), elbos_lin, color='blue')

fig3 = figure(
    title='KLD Progress',
    width=400,
    height=400,
)
fig3.line(range(len(klds_lin)), klds_lin, color='red')

gp = gridplot([[fig1], [fig2, fig3]])
show(gp)
```
