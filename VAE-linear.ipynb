```python
import numpy as np
from bokeh.io import show, output_notebook
from bokeh.plotting import figure
from torch import optim

from plotting import plot_model_outputs
from training import run_training, History
from vae import VAE

output_notebook()
```

```python
mean = np.array([3., 3.])
cov = np.array([[2., 1.5], [1.5, 2.]])
data = np.random.multivariate_normal(mean, cov, size=100)

fig = figure(
    title='Linear subspace',
    width=400,
    height=400,
    x_range=(-5, 10),
    y_range=(-5, 10),
)
fig.circle(data[:, 0], data[:, 1])
show(fig)
```

```python
model = VAE(latent_dim=2, nb_hidden=2)
optimizer = optim.Adam(model.parameters(), lr=1e-3, weight_decay=1e-6)
history = History()
```

```python
run_training(data, model, optimizer, 5000, history, log_interval=250, batch_size=20)
```

```python
box = ((-3,-3), (8,8))
plot_model_outputs(model, data, box)
history.plot()
```
