```python
%load_ext autoreload
%autoreload 2

import numpy as np
from bokeh.io import show, output_notebook
from bokeh.plotting import figure
from torch import optim

from plotting import plot_model_outputs
from training import run_training, History
from vae import VAE

output_notebook()
```

```python
mean = np.array([3., 2.])
cov = np.array([[2., 0.], [0., .2]])
data = np.random.multivariate_normal(mean, cov, size=100)
data[:, 1] = data[:, 1] + (data[:, 0]-mean[0])**2

fig = figure(
    title='Non-Linear subspace',
    width=400,
    height=400,
    x_range=(-5, 10),
    y_range=(-5, 10),
)
fig.circle(data[:, 0], data[:, 1])
x = np.linspace(-1, 7, 1000)
fig.line(x, (x-3)**2+2, color='red')
show(fig)
```

```python
model = VAE(latent_dim=2, nb_hidden=40)
optim = optim.Adam(model.parameters(), lr=1e-3, weight_decay=1e-6)
history = History()
```

```python
run_training(data, model, optim, 5000, history, batch_size=20, log_interval=250)
```

```python
plot_model_outputs(model, data)
```
