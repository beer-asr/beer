```python
import numpy as np
from bokeh.io import show, output_notebook
from bokeh.plotting import figure, gridplot

output_notebook()
```

```python
BATCH_SIZE = 20
```

```python
mean = np.array([3., 2.])
cov = np.array([[2., 0.], [0., .2]])
#x = np.linspace(-5, 5, 1000)
data = np.random.multivariate_normal(mean, cov, size=100)
data[:, 1] = data[:, 1] + (data[:, 0]-mean[0])**2

fig = figure(
    title='Non-Linear subspace',
    width=400,
    height=400,
    x_range=(-5, 10),
    y_range=(-5, 10),
)
fig.circle(data[:, 0], data[:, 1])
x = np.linspace(-1, 7, 1000)
fig.line(x, (x-3)**2+2, color='red')
show(fig)

data_nonlin = data
```

```python
%load_ext autoreload
%autoreload 2
from vae_lib import plot_loss, loss_function, VAE, run_training, History
```

```python
from torch import optim

model_nonlin = VAE(latent_dim=2, nb_hidden=40)
optim_nonlin = optim.Adam(model_nonlin.parameters(), lr=1e-3, weight_decay=1e-6)
history = History()
```

```python
run_training(data_nonlin, model_nonlin, optim_nonlin, 5000, history, BATCH_SIZE, log_interval=250)
```

```python
nonlin_box =((-0.1, 0.8), (6.1,10.1))
fig1 = figure(
    title='p(x)',
    width=400,
    height=400,
    x_range=(nonlin_box[0][0], nonlin_box[1][0]),
    y_range=(nonlin_box[0][1], nonlin_box[1][1]),
)
plot_loss(model_nonlin, fig1, data_nonlin, nb_samples=10, box=nonlin_box, resolution=200)
show(fig1)

history.plot()
```
